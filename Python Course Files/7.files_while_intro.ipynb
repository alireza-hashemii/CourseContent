# Files - an introduction to text files

## Different File Modes

| Mode | Name             | Description                          | File Exists?     | Pointer Position |
| ---: | ---------------- | ------------------------------------ | ---------------- | ---------------- |
|  `r` | Read             | Opens file for reading only          | ✅ Required       | Beginning        |
|  `w` | Write            | Opens file for writing (overwrites)  | ❌ Created if not | Beginning        |
|  `a` | Append           | Opens file for writing (adds to end) | ❌ Created if not | End              |
|  `x` | Exclusive Create | Creates a new file, fails if exists  | ❌ Must not exist | Beginning        |
|  `b` | Binary           | Binary mode (used with other modes)  | —                | —                |
|  `t` | Text             | Text mode (default)                  | —                | —                |
|  `+` | Update           | Read and write mode                  | ✅ Required       | Beginning        |


| Mode  | Meaning                          |
| ----- | -------------------------------- |
| `r+`  | Read and write                   | 
| `w+`  | Write and read (overwrites file) |
| `a+`  | Append and read                  |
| `rb`  | Read binary                      |
| `wb`  | Write binary                     |
| `ab`  | Append binary                    |
| `rb+` | Read/write binary                |


💡 Text mode (t) is default, so it’s usually omitted.<br>
💡 Binary mode (b) is used for files like images, audio, or videos.<br>
💡 The + symbol allows both reading and writing.<br>

# Questions

**Question 1: Read all the data stored in a text file.**

```python
with open("data.txt", "r") as file:
    data = file.read()

print(data)
```

**Question 2: Express how many lines of data is stored in a file.**

```python
with open("data.txt", "r") as file:
    data = file.readlines()

print(f'Found {len(data)} Lines of Data')
```

**Question 3: Express how many words is stored in a file.**

```python
with open("data.txt", "r") as file:
    data = file.read()
    words = data.split()

print(f'Found {len(words)} words totally')
```

**Question 4: Search for a specific word in a file.**

```python
word = 'president'

with open("data.txt", "r") as file:
    data = file.read()
    occurences = data.count(word)

print(f'Found the word: {word} for {occurences} times')
```
